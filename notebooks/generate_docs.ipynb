# Generate the HTML API documentation

Clones the repository, installs it and uses `pdoc` to build the HTML documentation of `src/rasher_qc` into `docs/`. Then zips the `docs` folder so it can be downloaded and uploaded to GitHub.

No GPU, dataset or Roboflow API key is needed. A CPU runtime is fine.

**Before running:** set `REPO_URL` in the first code cell to your repository.

```python
# 1. Clone the repository (edit REPO_URL)
import os
REPO_URL = "https://github.com/ElijahKhoury1/YOUR-REPO-NAME"
REPO_DIR = REPO_URL.rstrip("/").split("/")[-1].replace(".git", "")

os.chdir("/content")
if not os.path.isdir(REPO_DIR):
    !git clone {REPO_URL}
os.chdir(f"/content/{REPO_DIR}")
print(os.getcwd())
!ls
```

```python
# 2. Install the package and pdoc (pdoc imports the modules to document them)
!pip install -q -r requirements.txt
!pip install -q -e .
!pip install -q pdoc
```

```python
# 3. Generate the HTML docs into docs/
!rm -rf docs
!pdoc -o docs src/rasher_qc
!ls -R docs
```

```python
# 4. Zip and download the docs folder, then upload it to your GitHub repo
!zip -rq docs.zip docs
from google.colab import files
files.download("docs.zip")
```
